Dataset description

For this lab I used a dataset of Pokémon, stored in pokemon.csv. . Each row represents a single Pokémon species and includes information about its stats, typing, abilities, and other attributes. The dataset has numerical features such as attack, defense, speed, hp, weight_kg, and generation, as well as categorical features like type1 and type2. This dataset is appropriate for exploratory data analysis because it contains a mix of numeric and categorical variables and a clear target feature that can be used for modeling.

Target feature

The target feature for this project is is_legendary. This feature is binary, where 1 indicates that the Pokémon is legendary and 0 indicates that it is not legendary. The goal of the analysis is to understand how other features relate to whether a Pokémon is legendary.

Features used in this analysis

attack: A numeric stat representing the physical attack strength of the Pokémon. Values range roughly from very low (around 10–20) to very high (over 150).
defense: A numeric stat representing the physical defense of the Pokémon. Values range from low (around 20–30) to high (over 150).
speed: A numeric stat representing how fast the Pokémon is in battle. Values range from very slow (around 20–30) to very fast (over 140).
hp: A numeric stat representing the hit points or health of the Pokémon. Values range from low (around 20–30) to very high (over 200 for some special cases).
weight_kg: A numeric feature representing the weight of the Pokémon in kilograms. Values range from very light (less than 1 kg) to extremely heavy (hundreds of kilograms).
generation: A numeric feature from 1 to 6 indicating the game generation in which the Pokémon was introduced.
type1: A categorical feature representing the primary elemental type of the Pokémon, such as grass, fire, water, electric, psychic, normal, and many others.

These features were chosen because they are intuitive, well documented, and likely to have meaningful relationships with whether a Pokémon is legendary.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("pokemon.csv")

df = df[["attack", "defense", "speed", "hp", "weight_kg", "generation", "type1", "is_legendary"]]

df = df.dropna(subset=["weight_kg"])

df["is_legendary"] = df["is_legendary"].astype(int)

df.head()
```

Visualization 1: Attack vs legendary status

This visualization compares the distribution of attack values for legendary and non legendary Pokémon. A boxplot is used to show the median, spread, and potential outliers for each group.

Interpretation

Legendary Pokémon tend to have higher attack values than non legendary Pokémon. The median attack for legendary Pokémon is noticeably higher, and the overall distribution is shifted upward. This suggests that high attack is one of the characteristics that distinguishes legendary Pokémon from regular ones, although there is still some overlap between the two groups.

```python
plt.figure(figsize=(8,5))
sns.boxplot(x="is_legendary", y="attack", data=df)
plt.xticks([0,1], ["Non-legendary", "Legendary"])
plt.title("Attack by Legendary Status")
plt.xlabel("Legendary Status")
plt.ylabel("Attack")
plt.show()
```

Visualization 2: Speed vs legendary status

This visualization shows the distribution of speed values for legendary and non legendary Pokémon using a boxplot.

Interpretation

Legendary Pokémon generally have higher speed than non legendary Pokémon. The median speed for legendary Pokémon is higher, and the upper range of their speeds extends into very fast values that most non legendary Pokémon do not reach. This indicates that being fast is another common trait among legendary Pokémon, which may reflect their special status in battles.

```python
plt.figure(figsize=(8,5))
sns.boxplot(x="is_legendary", y="speed", data=df)
plt.xticks([0,1], ["Non-legendary", "Legendary"])
plt.title("Speed by Legendary Status")
plt.xlabel("Legendary Status")
plt.ylabel("Speed")
plt.show()
```

Visualization 3: Defense vs legendary status

This visualization compares the defense stat for legendary and non legendary Pokémon using a boxplot.

Interpretation

Legendary Pokémon also tend to have higher defense values than non legendary Pokémon. The median defense for legendary Pokémon is higher, and the distribution shows that many legendary Pokémon have strong defensive stats. However, the overlap between the two groups is more noticeable here than for attack and speed, suggesting that while defense is important, it may not be as strong a differentiator as attack or speed.

```python
plt.figure(figsize=(8,5))
sns.boxplot(x="is_legendary", y="defense", data=df)
plt.xticks([0,1], ["Non-legendary", "Legendary"])
plt.title("Defense by Legendary Status")
plt.xlabel("Legendary Status")
plt.ylabel("Defense")
plt.show()
```

Visualization 4: HP vs legendary status

This visualization shows the distribution of hp values for legendary and non legendary Pokémon using a boxplot.

Interpretation

Legendary Pokémon tend to have slightly higher hp than non legendary Pokémon, but the difference is less dramatic than for attack and speed. The medians are closer together, and there is substantial overlap in the distributions. This suggests that while some legendary Pokémon have very high hp, health alone is not a strong indicator of legendary status compared to offensive stats.

```python
plt.figure(figsize=(8,5))
sns.boxplot(x="is_legendary", y="hp", data=df)
plt.xticks([0,1], ["Non-legendary", "Legendary"])
plt.title("HP by Legendary Status")
plt.xlabel("Legendary Status")
plt.ylabel("HP")
plt.show()
```

Visualization 5: Weight vs legendary status

This visualization compares the distribution of weight_kg for legendary and non legendary Pokémon using a boxplot.

Interpretation

Legendary Pokémon show a wider spread in weight, with some extremely heavy species, but the median weight is not dramatically different from that of non legendary Pokémon. Both groups contain light and heavy Pokémon. This suggests that weight_kg does not have a strong or consistent relationship with legendary status, and it is not a reliable predictor on its own.

```python
plt.figure(figsize=(8,5))
sns.boxplot(x="is_legendary", y="weight_kg", data=df)
plt.xticks([0,1], ["Non-legendary", "Legendary"])
plt.title("Weight by Legendary Status")
plt.xlabel("Legendary Status")
plt.ylabel("Weight (kg)")
plt.yscale("log")
plt.show()
```

Visualization 6: Type distribution by legendary status

This visualization shows how primary types (type1) are distributed between legendary and non legendary Pokémon. A countplot with hue is used to compare the counts of each type for the two groups.

Interpretation

Certain types appear more frequently among legendary Pokémon than others. For example, types such as psychic, dragon, and steel are relatively common among legendary Pokémon, while some common types like normal and bug are rarely legendary. This indicates that elemental type is related to legendary status, with some types being favored for legendary designs.

```python
plt.figure(figsize=(12,6))
sns.countplot(data=df, x="type1", hue="is_legendary")
plt.xticks(rotation=90)
plt.legend(title="Legendary", labels=["Non-legendary", "Legendary"])
plt.title("Primary Type by Legendary Status")
plt.xlabel("Primary Type")
plt.ylabel("Count")
plt.show()
```

Visualization 7: Generation vs legendary status

This visualization shows how legendary and non legendary Pokémon are distributed across generations using a countplot.

Interpretation

Some generations contain more legendary Pokémon than others. Later generations often introduce multiple legendary Pokémon, so the proportion of legendary Pokémon can vary by generation. This suggests that generation is related to legendary status, but this relationship is driven more by game design choices than by inherent statistical properties of the Pokémon.

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df, x="generation", hue="is_legendary")
plt.legend(title="Legendary", labels=["Non-legendary", "Legendary"])
plt.title("Generation by Legendary Status")
plt.xlabel("Generation")
plt.ylabel("Count")
plt.show()
```

Big picture conclusion

From these visualizations, several features show clear relationships with the target feature is_legendary. Attack and speed are strongly related to legendary status, with legendary Pokémon having noticeably higher values on average. Defense and hp also tend to be higher for legendary Pokémon, but with more overlap between groups. Weight_kg does not show a strong or consistent relationship with legendary status, indicating that physical size is not a reliable indicator. The type1 and generation visualizations reveal that certain types and generations are more likely to include legendary Pokémon, reflecting design patterns in the games. Overall, offensive stats and certain elemental types appear to be the most related to legendary status, while weight and hp are less influential.
